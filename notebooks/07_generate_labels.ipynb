```python
import pandas as pd

df = pd.read_csv(
    "../data/processed/cleaned_patient_dataset.csv"
)

df.head()
```

```python
def assign_meal_plan(row):

    # Diabetes
    if row["Chronic_Disease"] == "Diabetes":
        return "Low-Carb Diet"

    # Heart Disease
    elif row["Chronic_Disease"] == "Heart Disease":
        return "Low-Fat Diet"

    # Obesity
    elif row["BMI"] >= 30:
        return "Low-Carb Diet"

    # High Cholesterol
    elif row["Cholesterol_Level"] >= 220:
        return "Low-Fat Diet"

    # Highly Active
    elif row["Exercise_Frequency"] >= 5:
        return "High-Protein Diet"

    # Active Lifestyle
    elif row["Daily_Steps"] >= 10000:
        return "High-Protein Diet"

    # Default
    else:
        return "Balanced Diet"
```

```python
df["Recommended_Meal_Plan"] = df.apply(
    assign_meal_plan,
    axis=1
)
```

```python
print(df["Recommended_Meal_Plan"].value_counts())
```

```python
df.to_csv(
    "../data/processed/cleaned_patient_dataset.csv",
    index=False
)

print("Dataset Updated Successfully")
```
