# Notebook 02 – Data Cleaning

## Objective

The purpose of this notebook is to prepare the patient dataset for machine learning by performing data cleaning and preprocessing.

### Tasks Performed

- Load the dataset
- Assess data quality
- Handle missing values
- Remove unnecessary columns
- Standardize categorical values
- Verify the cleaned dataset
- Save the cleaned dataset for further analysis

### Input

`data/raw/Personalized_Diet_Recommendations.csv`

### Output

`data/processed/cleaned_patient_dataset.csv`

## Step 1 – Import Required Libraries

```python
# ==========================================
# Import Required Libraries
# ==========================================

import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

print("Libraries imported successfully.")
```

## Step 2 – Load the Raw Dataset

```python
# ==========================================
# Load Dataset
# ==========================================

df = pd.read_csv("../data/raw/Personalized_Diet_Recommendations.csv")

print("Dataset Loaded Successfully!")

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

df.head()
```

## Step 3 – Data Quality Assessment

Before cleaning the data, we examine:

- Missing values
- Duplicate rows
- Data types
- Dataset structure

This helps determine the appropriate preprocessing strategy.

```python
# ==========================================
# Missing Values
# ==========================================

missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage (%)": round(df.isnull().sum()/len(df)*100,2)
})

missing = missing[missing["Missing Values"]>0]

missing.sort_values("Percentage (%)",ascending=False)
```

```python
# ==========================================
# Duplicate Records
# ==========================================

duplicates = df.duplicated().sum()

print(f"Duplicate Records : {duplicates}")
```

```python
# ==========================================
# Data Types
# ==========================================

df.dtypes
```

## Step 4 – Create a Working Copy

The original dataset is preserved by creating a copy before applying any transformations.

```python
clean_df = df.copy()

print("Working copy created successfully.")
```

## Step 5 – Remove Non-Predictive Features

The `Patient_ID` column is a unique identifier and does not contribute to prediction. It is removed before model training.

```python
clean_df.drop(columns=["Patient_ID"], inplace=True)

print("Patient_ID removed.")

print(clean_df.shape)
```

## Step 6 – Handle Missing Values

The missing values represent the absence of a condition rather than missing information.

Therefore:

- Chronic_Disease → "No Disease"
- Allergies → "None"
- Food_Aversions → "None"

```python
clean_df["Chronic_Disease"] = clean_df["Chronic_Disease"].fillna("No Disease")

clean_df["Allergies"] = clean_df["Allergies"].fillna("None")

clean_df["Food_Aversions"] = clean_df["Food_Aversions"].fillna("None")
```

```python
clean_df.isnull().sum()
```

## Step 7 – Standardize Categorical Values

Leading and trailing spaces are removed from all categorical features to ensure consistency.

```python
categorical_columns = clean_df.select_dtypes(include="object").columns

for col in categorical_columns:
    clean_df[col] = clean_df[col].str.strip()
```

## Step 8 – Verify the Cleaned Dataset

```python
clean_df.info()
```

```python
print(clean_df.head())
print(clean_df.shape)
```

## Step 9 – Save the Cleaned Dataset

```python
clean_df.to_csv(
    "../data/processed/cleaned_patient_dataset.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```

# Conclusion

The dataset has been successfully cleaned and prepared for further analysis.
